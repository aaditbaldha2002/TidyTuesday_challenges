##### Checking if the company_id is unique or not to figure out the duplicate rows

```python
company_df.shape[0]==len(company_df['company_id'].unique())
```

```python
print(company_df['legal_nature'].unique())
legal_nature_counts = company_df['legal_nature'].value_counts()
print(legal_nature_counts)
plt.figure(figsize=(10,6))
legal_nature_counts.plot(kind='bar')
plt.title('Distribution of Legal Nature of Companies')
plt.xlabel('Legal Nature')
plt.ylabel('Number of Companies')
plt.show()
```

```python
company_df['owner_qualification'].unique()
owner_qualification_counts = company_df['owner_qualification'].value_counts()
print(owner_qualification_counts)
plt.figure(figsize=(10,6))
owner_qualification_counts.plot(kind='bar')
plt.title('Distribution of Owner Qualifications')
plt.xlabel('Owner Qualification')
plt.ylabel('Number of Companies')
plt.show()
```

```python
company_df['company_size'].unique()
company_size_counts = company_df['company_size'].value_counts()
print(company_size_counts)
plt.figure(figsize=(10,6))
company_size_counts.plot(kind='bar')
plt.title('Distribution of Company Sizes')
plt.xlabel('Company Size')
plt.ylabel('Number of Companies')
plt.show()
```
