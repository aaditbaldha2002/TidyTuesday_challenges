```python
legal_nature_df=pd.read_csv("data/legal_nature.csv")
print(legal_nature_df)
legal_nature_df.shape
```

```python
legal_nature_df=legal_nature_df.reset_index().drop(columns=['id']).rename(columns={"index":"id"})
print(legal_nature_df)
```

```python
qualifications_df=pd.read_csv("data/qualifications.csv")
print(qualifications_df)
qualifications_df.shape
```

```python
qualifications_df=qualifications_df.reset_index().drop(columns=['id']).rename(columns={"index":"id"})
print(qualifications_df)
```

```python
size_df=pd.read_csv("data/size.csv")
print(size_df)
size_df.shape
```

```python
size_df=size_df.reset_index().drop(columns=['id']).rename(columns={"index":"id"})
print(size_df)
```

```python
company_ln_df=pd.merge(company_df, legal_nature_df, how='left',on='legal_nature').rename(columns={"id":"legal_nature_id"}).drop(columns=['legal_nature'])
print(company_ln_df.head(10))
```

```python
company_oc_df=pd.merge(company_ln_df, qualifications_df, how='left',on='owner_qualification').rename(columns={"id":"owner_qualification_id"}).drop(columns=['owner_qualification'])
print(company_oc_df.head(10))
```

```python
total_company_df=pd.merge(company_oc_df, size_df, how='left',on='company_size').rename(columns={"id":"company_size_id"}).drop(columns=['company_size'])
print(total_company_df.head(10))
```

##### Which legal nature categories concentrate the highest total and average capital stock?

```python
result = (
    total_company_df.groupby("legal_nature_id")["capital_stock"]
      .agg(avg="mean", max="max")
      .sort_values("avg", ascending=False)
)

plt.figure(figsize=(14, 10))
plt.bar(result.index, result["avg"])
plt.title("Average Company Stock Price by Legal Nature")
plt.ylabel("Average Stock Price")
plt.xlabel("Legal Nature")
plt.show()
```

```python
import numpy as np

# Number of buckets = number of unique legal_nature values
x = np.arange(result.index.nunique())

plt.figure(figsize=(8, 5))

plt.bar(
    x,
    result["avg"],
    label="Average Stock Price"
)

plt.scatter(
    x,
    result["max"],
    label="Maximum Stock Price",
    zorder=3
)

plt.yscale("log")
plt.xticks(x, result.index)  # one tick per unique legal_nature
plt.title("Average vs Max Company Stock Price by Legal Nature (Log Scale)")
plt.ylabel("Stock Price (log scale)")
plt.xlabel("Legal Nature")
plt.legend()
plt.show()
```

##### How does company size relate to capital stock (and how skewed is it)?

```python
result= total_company_df.groupby("company_size_id")["capital_stock"].agg(avg="mean", max="max")
result
```

```python
plt.figure(figsize=(8, 5))

bars = plt.bar(
    result.index,
    result["avg"],
    label="Average Capital Stock"
)

points = plt.scatter(
    result.index,
    result["max"],
    label="Maximum Capital Stock",
    zorder=3,
    color='red'
)

plt.yscale("log")
plt.xticks(result.index)
plt.title("Average vs Max Capital Stock by Company Size (Log Scale)")
plt.ylabel("Capital Stock (log scale)")
plt.xlabel("Company Size")

# Annotate bars with offset above
for bar in bars:
    height = bar.get_height()
    plt.text(
        bar.get_x() + bar.get_width() / 2,
        height * 1.1,  # small offset above bar
        f"{height:,.0f}",
        ha="center",
        va="bottom",
        fontsize=9
    )

# Annotate scatter points
for x, y in zip(result.index, result["max"]):
    plt.text(
        x,
        y * 1.05,
        f"{y:,.0f}",
        ha="center",
        va="bottom",
        fontsize=9
    )

plt.legend()
plt.show()
```

##### Do specific owner qualification groups dominate high-capital companies?

```python
result = total_company_df.groupby("owner_qualification_id")["capital_stock"].agg(avg="mean")
result
```

```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

# Step 1: Sort by descending average
result_sorted = result.sort_values(by="avg", ascending=False)

# Step 2: Create numeric buckets for x-axis
x = np.arange(result_sorted.index.nunique())

# Step 3: Plot the bar chart
plt.figure(figsize=(10, 6))
bars = plt.bar(
    x,
    result_sorted["avg"],
    label="Average Capital Stock",
    color='skyblue'
)

# Step 4: Label the bars with the actual numbers
for bar in bars:
    height = bar.get_height()
    plt.text(
        bar.get_x() + bar.get_width() / 2,
        height * 1.01,  # slightly above the bar
        f"{height:,.0f}",
        ha="center",
        va="bottom",
        fontsize=9
    )

# Step 5: Set x-ticks and labels
plt.xticks(x, result_sorted.index)

# Step 6: Labels, title, and legend
plt.title("Average Capital Stock by Owner Qualification (Descending)")
plt.xlabel("Owner Qualification")
plt.ylabel("Average Capital Stock")
plt.legend()
plt.tight_layout()  # Prevent clipping of x-tick labels
plt.show()
```
